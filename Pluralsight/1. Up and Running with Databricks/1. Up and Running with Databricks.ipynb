# 1. Analysing data using notebook

**dbutils.fs.ls("/databricks-datasets"):**

dbutils is Databricks' built-in utility library for things like file system operations, secrets, and notebook workflows.

- `.fs `is the file system module within dbutils.

- `.ls(path)` lists the files and directories at the given path — just like running ls in a terminal.

So this specific command lists everything inside the /databricks-datasets folder, which is a special directory Databricks provides out-of-the-box with a bunch of sample datasets (like the airlines dataset mentioned in the course demo) you can use for practice and experimentation.

It'll return a list of file objects, each showing the path, name, size, and modification time of items in that folder.

```python
dbutils.fs.ls("/databricks-datasets")
```

```python
    # Storing the existing csv file /databricks-datasets/airlines/part-00000 in df

file_path = "/databricks-datasets/airlines/part-00000"
df = spark.read.csv(file_path, header =True, inferSchema=True)
df.display()
```

```python
# Print df results
display(df)
```

```python
# Filtering the data
df_delay = df.filter(df["IsDepDelayed"]=="YES")
df_delay.display()
```

```python
# Select few columns
selected_df = df.select('Year', 'Month', 'DayofMonth', 'Origin', 'Dest')

# Display first 10 rows
selected_df.limit(10).display()
```

```python
from pyspark.sql.functions import col

# Creating a new column called TOtal Delay which sum the values from ArrDelay and DeDelay columns
from pyspark.sql.functions import col, expr

# Creating a new column called TotalDelay which sums the values from ArrDelay and DepDelay columns
# try_cast used since there are some invalid values while trying to convert values to number
df_with_total_delay = df.limit(1000).withColumn("TotalDelay", expr("try_cast(ArrDelay as int)") + expr("try_cast(DepDelay as int)") )

# Display first 5000 rows
display(df_with_total_delay.limit(5000))
```
