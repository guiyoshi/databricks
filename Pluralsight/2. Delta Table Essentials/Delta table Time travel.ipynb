# Delta table Time travel

```python
spark.sql("USE silver_wwi")
```

```python
%sql
DESCRIBE DETAIL workspace.silver_wwi.stockitems
```

```python
%sql
DESCRIBE HISTORY workspace.silver_wwi.stockitems
```

**View current table then timestamp**

```python
%sql
SELECT * FROM silver_wwi.stockitems
WHERE ColorID IS NULL
```

**Query data from table as of a timestamp**

```python
%sql
SELECT * FROM silver_wwi.stockitems
TIMESTAMP AS OF '2026-10-09T00:17:20.000+00:00'
WHERE ColorID IS NULL
```

**Restore previous version**

```python
%sql
DELETE FROM silver_wwi.stockitems
WHERE ColorID = 0
```

```python
%sql
DESCRIBE HISTORY silver_wwi.stockitems
```

```python
%sql
RESTORE TABLE silver_wwi.stockitems TO VERSION AS OF 2
```

```python
%sql
DESCRIBE HISTORY silver_wwi.stockitems
```

```python
%sql
SELECT * FROM silver_wwi.stockitems
WHERE ColorID = 0
```
