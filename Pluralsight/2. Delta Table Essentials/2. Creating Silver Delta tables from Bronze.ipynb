# Creating Silver Delta tables from Bronze

<b> Make location the Bronze WWI database/warehouse <b>

```python
spark.sql("use bronze_wwi")
```

<b>SQL Describe the History of Delta table<b>

```python
%sql
DESCRIBE HISTORY workspace.bronze_wwi.stockitems
```

<b> Write data to table <b>

```python
dfStockItems = spark.sql("select * from workspace.bronze_wwi.stockitems")
dfStockItems.display()
```

```python
# Setting up database and table

spark.sql("create database if not exists silver_wwi")
spark.sql("use silver_wwi")
spark.sql("drop table if exists stockitems")
```

```python
# Materializing the delta table in overwrite mode
dfStockItems.write.format("delta").mode("overwrite").saveAsTable("stockitems")
```

```python
%sql
DESCRIBE HISTORY workspace.silver_wwi.stockitems
```

<b> Applying changes to the table and tracking history <b>

```python
%sql
-- Update ColorID to 0 where NULL
UPDATE workspace.silver_wwi.stockitems
    SET ColorID = 0 WHERE ColorID IS NULL
```

```python
dfsilverstockitems = spark.sql("select * from workspace.silver_wwi.stockitems")
dfsilverstockitems.display()
```
